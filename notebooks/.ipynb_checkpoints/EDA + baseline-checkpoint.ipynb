We can see that the summaries are more complex than the dialogues. What conclusions can we draw?
- Our task is not simply to condense the text, but to rewrite the dialogues from scratch using the appropriate vocabulary

# N-gramms

```python
vectorizer = CountVectorizer(ngram_range=(2, 2), max_features=20, stop_words='english')
X = vectorizer.fit_transform(train['dialogue'])
ngrams = vectorizer.get_feature_names_out()
freqs = X.sum(axis=0).A1

plt.figure(figsize=(10,5))
sns.barplot(x=sorted(freqs), y=ngrams)
plt.title("Top Bigrams in Dialogues",fontsize=25,pad=20,fontweight='bold',color='darkcyan')
plt.tight_layout()
plt.show()
```

```python
vectorizer = CountVectorizer(ngram_range=(2, 2), max_features=20, stop_words='english')
X = vectorizer.fit_transform(train['summary'])
ngrams = vectorizer.get_feature_names_out()
freqs = X.sum(axis=0).A1

plt.figure(figsize=(10,5))
sns.barplot(x=sorted(freqs), y=ngrams)
plt.title("Top Bigrams in Summaries",fontsize=25,pad=20,fontweight='bold',color='darkcyan')
plt.tight_layout()
plt.show()
```

The most popular bigrams in the dialogues and summaries differ, which may indicate that the summaries are text rewritten from scratch. Let's check this using the novel n-gram ratio.

## Novel n-gram ratio

```python
stemmer = SnowballStemmer("english")

@lru_cache(maxsize=None)
def stem(word):
    return stemmer.stem(word)

def tokenize(text, use_stem=True):
    tokens = re.findall(r"\w+", text.lower())
    return [stem(t) for t in tokens] if use_stem else tokens

def get_ngrams(tokens, n):
    return [tuple(tokens[i:i+n]) for i in range(len(tokens) - n + 1)]

def novel_ratio(dialogue, summary, n, use_stem):
    d = set(get_ngrams(tokenize(dialogue), n))
    s = set(get_ngrams(tokenize(summary), n))
    if not s:
        return None
    return sum(g not in d for g in s) / len(s)

def compute(df, dialogue_col="dialogue", summary_col="summary"):
    rows = {}
    for use_stem in (False, True):
        name = "stem" if use_stem else "raw"
        rows[name] = {
            f"{n}-gram": pd.Series(
                [novel_ratio(d, s, n, use_stem)
                 for d, s in zip(df[dialogue_col], df[summary_col])]
            ).mean()
            for n in (1, 2, 3, 4)
        }
    return pd.DataFrame(rows).T

print(compute(test))
```

The metric values are very high starting with 2-grams, which means that the summary is indeed text rewritten from scratch

## Are there any conversations in the dataset involving more than two people?

```python
def get_speakers(dialogue):
    return set(re.findall(r'#Person\d+#', dialogue))

train['speakers'] = train['dialogue'].apply(get_speakers)
train['num_speakers'] = train['speakers'].apply(len)

print(train['num_speakers'].value_counts().sort_index())
```

Let's check to see if everything is OK with dialogs involving a large number of people

# Baseline (first + last sentence)

```python
train['dialogue'].iloc[0]
```

```python
def baseline(dialogue):
    sents = sent_tokenize(dialogue.strip())
    return sents[0] + ' ' + sents[-1]
```

```python
test['baseline'] = test['dialogue'].apply(baseline)
```

## ROUGE and BLEU scores

### BLEU

```python
test['dialog_id'] = test['id'].str.rsplit('_', n=1).str[0]

g = test.groupby('dialog_id', sort=False)
hyps = g['baseline'].first().astype(str).tolist()
refs_lists = g['summary'].apply(list).tolist()

n_refs = max(len(r) for r in refs_lists)
padded = [r + [r[-1]] * (n_refs - len(r)) for r in refs_lists]
refs = [[r[i] for r in padded] for i in range(n_refs)]

bleu = sacrebleu.corpus_bleu(hyps, refs)
print(bleu.score)
```

```python
scorer = rouge_scorer.RougeScorer(["rouge1", "rouge2", "rougeL"], use_stemmer=True)

res = {k: [] for k in ["rouge1", "rouge2", "rougeL"]}
for hyp, rs in zip(hyps, refs_lists):
    scores = [scorer.score(r, hyp) for r in rs]
    for k in res:
        res[k].append(max(s[k].fmeasure for s in scores))

for k, v in res.items():
    print(k, np.mean(v))
```

The scores are low, which is expected for a baseline. Trained models on DialogSum usually reach ROUGE-1 around 45-47 and ROUGE-2 around 20. If the baseline is just a chunk of the original dialogue, the moderate ROUGE-1 comes from shared words (names, keywords), while the very low ROUGE-2 shows that the reference summaries paraphrase rather than copy the dialogue lines. These numbers set a lower bound to compare your model against.
